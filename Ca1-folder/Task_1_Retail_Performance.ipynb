## 7. Data Preparation and Quality Assessment

This section checks the quality of the raw dataset before analysis. The original `df` is kept unchanged, while a separate `df_clean` will be prepared for later analysis.

### 7.1 Missing-value assessment

This check counts missing values in each column. It is run before imputation so the original data quality issue can be reported.

```python
# Count missing values in each column.
missing_values = df.isnull().sum()
missing_percentage = (missing_values / len(df)) * 100

# Combine the counts and percentages into one easy-to-read summary.
missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage
})

missing_summary
```

### 7.2 Duplicate-record assessment

This check identifies completely duplicated rows. Duplicate rows can cause some stores or observations to be counted more than once.

```python
# Count completely duplicated rows.
duplicate_count = df.duplicated().sum()
print(f"Number of completely duplicated rows: {duplicate_count}")

# Display duplicated rows, if any exist.
duplicated_rows = df[df.duplicated(keep=False)]
duplicated_rows
```

### 7.3 Handle duplicate records

Only completely identical rows are removed. Repeated `StoreID` values are retained because they may represent legitimate observations with different performance values.

```python
# Make a clean copy of the original dataset and remove exact duplicates.
df_clean = df.copy()
rows_before = len(df_clean)
df_clean = df_clean.drop_duplicates().copy()
rows_removed = rows_before - len(df_clean)

print(f"Completely duplicated rows removed: {rows_removed}")
print(f"Rows remaining: {len(df_clean)}")
```

### 7.4 Outlier assessment using the IQR method

This check flags unusually low or high numerical values using the interquartile range (IQR). Outliers are reported for review rather than automatically deleted, because unusual business performance may be genuine.

```python
# Select numerical business measures for outlier investigation.
# StoreID is an identifier, so it is excluded from this calculation.
numerical_columns = df.select_dtypes(include=np.number).columns
outlier_columns = numerical_columns.drop("StoreID", errors="ignore")

# Calculate the lower and upper IQR boundaries for each business measure.
q1 = df[outlier_columns].quantile(0.25)
q3 = df[outlier_columns].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

# Count the potential outliers in each business measure.
outlier_summary = pd.DataFrame({
    "Lower Bound": lower_bound,
    "Upper Bound": upper_bound,
    "Potential Outliers": ((df[outlier_columns] < lower_bound) | (df[outlier_columns] > upper_bound)).sum()
})

outlier_summary
```

### 7.5 Handle missing values

A separate `df_clean` DataFrame is created so the raw dataset remains available for comparison. Numerical missing values are replaced with their column median, while categorical missing values are replaced with the most common value.

```python
# Identify numerical and categorical columns in the prepared dataset.
numerical_columns = df_clean.select_dtypes(include=np.number).columns
categorical_columns = df_clean.select_dtypes(exclude=np.number).columns

# Replace missing numerical values with the median of each column.
for column in numerical_columns:
    df_clean[column] = df_clean[column].fillna(df_clean[column].median())

# Replace missing categorical values with the most common value in each column.
for column in categorical_columns:
    df_clean[column] = df_clean[column].fillna(df_clean[column].mode()[0])

print("Missing values after imputation:")
print(df_clean.isnull().sum())
```

### 7.6 Create engineered features

These features support later commercial analysis by combining existing measures into total traffic, sales change from the previous month, and sales generated per staff hour.

```python
# Feature 1: combine physical and online visitors.
df_clean["TotalTraffic"] = df_clean["Footfall"] + df_clean["OnlineTraffic"]

# Feature 2: calculate the change in sales compared with the previous month.
df_clean["SalesChange"] = df_clean["SalesRevenue"] - df_clean["PreviousMonthSales"]

# Feature 3: calculate sales revenue generated per staff hour.
df_clean["SalesPerStaffHour"] = df_clean["SalesRevenue"] / df_clean["StaffHours"]

# Display the new features for a quick check.
df_clean[["TotalTraffic", "SalesChange", "SalesPerStaffHour"]].head()
```

### 7.7 Final quality check

This final check confirms the prepared dataset dimensions, remaining missing values and duplicate rows before analysis continues.

```python
# Confirm the final shape of the prepared dataset.
print(f"Prepared dataset shape: {df_clean.shape}")

# Confirm that no missing values remain after imputation.
print(f"Total missing values remaining: {df_clean.isnull().sum().sum()}")

# Confirm that no complete duplicate rows remain.
print(f"Complete duplicate rows remaining: {df_clean.duplicated().sum()}")

# Display the prepared dataset structure.
df_clean.info()
```

## 8. Exploratory Data Analysis

This section explores the prepared dataset from a Commercial Director perspective. The analysis focuses on revenue performance, regional contribution, store formats, seasonality, customer traffic and marketing effectiveness.

### 8.1 Overall commercial performance

This summary provides key measures of scale and performance before comparing regions, stores, seasons and marketing-spend levels.

```python
# Summarise the main commercial performance measures.
overall_summary = pd.DataFrame({
    "Measure": [
        "Total Sales Revenue",
        "Average Sales Revenue",
        "Total Marketing Spend",
        "Average Total Traffic",
        "Average Customer Satisfaction"
    ],
    "Value": [
        df_clean["SalesRevenue"].sum(),
        df_clean["SalesRevenue"].mean(),
        df_clean["MarketingSpend"].sum(),
        df_clean["TotalTraffic"].mean(),
        df_clean["CustomerSatisfaction"].mean()
    ]
})

overall_summary
```

### 8.2 GroupBy analysis 1: regional performance

This analysis compares total revenue, average revenue, marketing spend and traffic across regions.

```python
# Compare commercial performance across regions.
regional_performance = (
    df_clean.groupby("Region")
    .agg(
        TotalSales=("SalesRevenue", "sum"),
        AverageSales=("SalesRevenue", "mean"),
        TotalMarketingSpend=("MarketingSpend", "sum"),
        AverageTraffic=("TotalTraffic", "mean")
    )
    .sort_values("TotalSales", ascending=False)
)

regional_performance
```

### 8.3 GroupBy analysis 2: store-format performance

This analysis compares Mall, Town and Retail Park formats to identify differences in revenue, traffic and customer satisfaction.

```python
# Compare commercial performance across store formats.
store_type_performance = (
    df_clean.groupby("StoreType")
    .agg(
        NumberOfObservations=("StoreID", "count"),
        TotalSales=("SalesRevenue", "sum"),
        AverageSales=("SalesRevenue", "mean"),
        AverageTraffic=("TotalTraffic", "mean"),
        AverageSatisfaction=("CustomerSatisfaction", "mean")
    )
    .sort_values("TotalSales", ascending=False)
)

store_type_performance
```

### 8.4 GroupBy analysis 3: seasonal performance

This analysis investigates how sales revenue, traffic and marketing spend vary by season.

```python
# Compare commercial performance across seasons.
seasonal_performance = (
    df_clean.groupby("Season")
    .agg(
        TotalSales=("SalesRevenue", "sum"),
        AverageSales=("SalesRevenue", "mean"),
        AverageMarketingSpend=("MarketingSpend", "mean"),
        AverageTraffic=("TotalTraffic", "mean")
    )
    .sort_values("TotalSales", ascending=False)
)

seasonal_performance
```

### 8.5 GroupBy analysis 4: marketing-spend performance

Marketing spend is divided into three equal-sized bands. This helps examine whether higher marketing investment is associated with higher sales and traffic.

```python
# Create three marketing-spend bands using the 33rd and 67th percentiles.
marketing_bands = [
    df_clean["MarketingSpend"].min() - 1,
    df_clean["MarketingSpend"].quantile(0.33),
    df_clean["MarketingSpend"].quantile(0.67),
    df_clean["MarketingSpend"].max()
]

band_labels = ["Low", "Medium", "High"]
df_clean["MarketingSpendBand"] = pd.cut(
    df_clean["MarketingSpend"],
    bins=marketing_bands,
    labels=band_labels,
    include_lowest=True
)

# Compare performance across marketing-spend bands.
marketing_performance = (
    df_clean.groupby("MarketingSpendBand", observed=True)
    .agg(
        NumberOfObservations=("StoreID", "count"),
        AverageMarketingSpend=("MarketingSpend", "mean"),
        AverageSales=("SalesRevenue", "mean"),
        AverageTraffic=("TotalTraffic", "mean")
    )
    .reindex(band_labels)
)

marketing_performance
```

### 8.6 Top-performing stores and regions

These rankings identify the largest contributors to revenue. Store rankings aggregate observations for the same `StoreID`, while regional rankings use the regional GroupBy analysis above.

```python
# Rank stores by total sales revenue.
top_5_stores = (
    df_clean.groupby("StoreID")
    .agg(
        TotalSales=("SalesRevenue", "sum"),
        AverageSales=("SalesRevenue", "mean"),
        NumberOfObservations=("SalesRevenue", "count")
    )
    .sort_values("TotalSales", ascending=False)
    .head(5)
)

# Rank regions by total sales revenue.
top_5_regions = regional_performance.head(5)

print("Top 5 stores by total sales:")
display(top_5_stores)

print("Top 5 regions by total sales:")
display(top_5_regions)
```

### 8.7 Initial commercial findings

This cell extracts the strongest-performing category from each comparison. These are analytical findings for Task 1; detailed recommendations and management actions belong in Task 3.

```python
# Identify the strongest category in each commercial comparison.
highest_region = regional_performance["TotalSales"].idxmax()
highest_store_type = store_type_performance["TotalSales"].idxmax()
highest_season = seasonal_performance["TotalSales"].idxmax()
highest_marketing_band = marketing_performance["AverageSales"].idxmax()

print(f"Highest total-sales region: {highest_region}")
print(f"Highest total-sales store format: {highest_store_type}")
print(f"Highest total-sales season: {highest_season}")
print(f"Highest average-sales marketing band: {highest_marketing_band}")
```
